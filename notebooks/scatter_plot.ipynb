```python
# IMPORT LIBRARIES
import pandas as pd
import matplotlib.pyplot as plt

# LOAD CSV
srd = pd.read_csv(
    "csv/Sangareddy_NDVI_By_Mandal_TOTAL_MCM_2021.csv"
)

# DEFINE VARIABLES
x = srd["Total_MCM"]

y = srd["NDVI"]

labels = srd["Mandal"]

# CREATE FIGURE
plt.figure(figsize=(16, 12))

# CREATE SCATTER PLOT
plt.scatter(
    x,
    y,
    alpha=0.7
)

# ADD MANDAL LABELS
for i, txt in enumerate(labels):

    plt.annotate(

        txt,

        (
            x[i] + 1,
            y[i] - 0.0001
        ),

        fontsize=7
    )

# X LABEL
plt.xlabel(
    "Agricultural Water Demand (MCM)"
)

# Y LABEL
plt.ylabel(
    "NDVI"
)

# TITLE
plt.title(
    "NDVI vs Agricultural Water Demand"
)

# GRID
plt.grid(True)

# SAVE FIGURE
plt.savefig(
    "maps/srd_ndvi_vs_totalMCM_2021_kharif.png",
    dpi=300,
    bbox_inches="tight"
)

# SHOW PLOT
plt.show()
```
