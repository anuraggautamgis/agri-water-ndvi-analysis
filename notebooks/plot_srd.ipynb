```python
# Library and Shapefile import
import geopandas as gpd
import matplotlib.pyplot as plt

srd = gpd.read_file("C:/Users/anura/Desktop/Sangareddy_Portfolio/Sangareddy_QGIS/sangareddy_mandal.shp")


# Plot Map
fig, ax = plt.subplots(
    figsize=(14,14)
)

srd.plot(ax=ax,
         edgecolor="black",
         facecolor="lightgreen"
         )

for idx, row in srd.iterrows():

    point = row.geometry.representative_point()

    plt.annotate(

        text=row["Mandal"],

        xy=(
            point.x,
            point.y
        ),

        horizontalalignment="center",

        fontsize=7
    )

# Plot Title
plt.title("Sangareddy Mandal Boundary Map")

# Axis
plt.axis("off")

# Save Map
plt.savefig("maps/sangareddy_mandal_boundary.png",
            dpi=300,
            bbox_inches="tight")

# Show Map
plt.show()
```
